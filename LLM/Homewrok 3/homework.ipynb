# Домашнее задание по визуализации данных (matplotlib)

## Задание 1: Зависимость веса от роста


```python
import pandas as pd
import matplotlib.pyplot as plt

# Читаю данные из CSV
df = pd.read_csv('rost_ves.csv', encoding='utf-8-sig')

# Создаю списки ростов и весов
rost = df['Рост (см)'].tolist()
ves = df['Вес (кг)'].tolist()

# Постройка точечного графика
plt.figure(figsize=(10, 6))
plt.scatter(rost, ves, color='green', s=100, alpha=0.7)

# Добавляю подписи осей и заголовок
plt.xlabel('Рост (см)', fontsize=12)
plt.ylabel('Вес (кг)', fontsize=12)
plt.title('Зависимость веса от роста', fontsize=14, fontweight='bold')

# Отображение сетки для удобства
plt.grid(True, alpha=0.3)

# Отображаю график
plt.tight_layout()
plt.show()
```

## Задание 2: Температура в течение суток


```python
# Создаю списки часов и температур
hours = [0, 4, 8, 12, 16, 20, 24]
temperature = [15, 14, 18, 22, 25, 20, 17]

# Постройка линейного графика
plt.figure(figsize=(10, 6))
plt.plot(hours, temperature, marker='o', linewidth=2, markersize=8, color='red', markerfacecolor='blue')

# Добавляю заголовок и подписи осей
plt.xlabel('Часы', fontsize=12)
plt.ylabel('Температура (°C)', fontsize=12)
plt.title('Температура в течение суток', fontsize=14, fontweight='bold')

# Включаю отображение сетки
plt.grid(True, alpha=0.3, linestyle='--')

# Отображаю результат
plt.tight_layout()
plt.show()
```
