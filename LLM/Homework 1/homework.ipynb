## Задание 1: Работа с двумерными массивами


```python
import numpy as np

# Создаю двумерный массив 4x4 со случайными числами от 1 до 50
arr = np.random.randint(1, 51, size=(4, 4))

# Вывожу массив в читаемом виде
print("Исходный массив 4x4:")
print(arr)
print()

# Суммы по строкам (axis=0 - это столбцы, axis=1 - строки, мать твою!)
sum_rows = np.sum(arr, axis=1)
print(f"Суммы по строкам: {sum_rows}")

# Суммы по столбцам
sum_cols = np.sum(arr, axis=0)
print(f"Суммы по столбцам: {sum_cols}")

# Общее среднее значение
mean_value = np.mean(arr)
print(f"Общее среднее значение: {mean_value:.2f}")
```

## Задание 2: Транспонирование и изменение формы массива


```python
# Создаю массив 3x5 с числами от 0 до 14
arr2 = np.arange(0, 15).reshape(3, 5)

print("Исходный массив 3x5:")
print(arr2)
print()

# Транспонирование (строки становятся столбцами, столбцы - строками)
arr2_transposed = arr2.T
print("Транспонированный массив (5x3):")
print(arr2_transposed)
print()

# Применяю reshape для получения 5x3
arr2_reshaped = arr2.reshape(5, 3)
print("Массив после reshape 5x3:")
print(arr2_reshaped)
```

# Отчёт по продажам

## В этом разделе мы считаем итоги по простым данным


```python
# Список продаж за неделю
sales = [120, 135, 100, 150, 90, 200, 170]

# Общая сумма
total_sum = np.sum(sales)

# Среднее значение
average = np.mean(sales)

# Максимум
maximum = np.max(sales)

# Минимум
minimum = np.min(sales)

print(f"Общая сумма: {total_sum}")
print(f"Среднее: {average:.2f}")
print(f"Максимум: {maximum}")
print(f"Минимум: {minimum}")
```
