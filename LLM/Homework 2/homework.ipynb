## Задание 1: Работа с DataFrame книг


```python
import pandas as pd
import numpy as np

# Создаю DataFrame с книгами
books_data = {
    'Название': ['Война и мир', 'Преступление и наказание', '1984', 'Мастер и Маргарита', 
                 'Три товарища', 'Собачье сердце', '451 градус по Фаренгейту', 'Анна Каренина'],
    'Автор': ['Лев Толстой', 'Фёдор Достоевский', 'Джордж Оруэлл', 'Михаил Булгаков',
              'Эрих Мария Ремарк', 'Михаил Булгаков', 'Рэй Брэдбери', 'Лев Толстой'],
    'Год издания': [1869, 1866, 2020, 1967, 2018, 2021, 1953, 2019],
    'Цена': [450, 320, 380, 410, 390, 280, 330, 450]
}

df_books = pd.DataFrame(books_data)

# Вывожу весь DataFrame
print("Исходный DataFrame с книгами:")
print(df_books)
print()

# Средняя цена
avg_price = df_books['Цена'].mean()
print(f"Средняя цена книг: {avg_price:.2f} руб.")
print()

# Книги после 2015 года
books_after_2015 = df_books[df_books['Год издания'] > 2015]
print("Книги, изданные после 2015 года:")
print(books_after_2015)
print()

# Сортировка по цене по возрастанию
df_books_sorted = df_books.sort_values('Цена', ascending=True)
print("DataFrame, отсортированный по цене (возрастание):")
print(df_books_sorted)
```

## Задание 2: Работа с CSV файлом заказов


```python
# Создаю CSV с данными о заказах
orders_data = {
    'Номер заказа': range(1, 26),
    'Клиент': ['Иванов', 'Петров', 'Сидоров', 'Иванов', 'Смирнов', 
               'Петров', 'Иванов', 'Козлов', 'Петров', 'Новиков',
               'Сидоров', 'Иванов', 'Смирнов', 'Петров', 'Иванов',
               'Козлов', 'Новиков', 'Петров', 'Иванов', 'Сидоров',
               'Смирнов', 'Петров', 'Иванов', 'Козлов', 'Новиков'],
    'Дата': pd.date_range('2024-01-01', periods=25, freq='D'),
    'Сумма': [1500, 2300, 1800, 3500, 1200, 4100, 2800, 1900, 3200, 2500,
              1600, 3700, 2200, 4400, 2100, 2900, 3300, 1800, 4500, 1700,
              3100, 2600, 3900, 2400, 4200]
}

df_orders = pd.DataFrame(orders_data)

# Сохраняю в CSV
df_orders.to_csv('orders.csv', index=False, encoding='utf-8-sig')

# Читаю данные из файла
df_orders_read = pd.read_csv('orders.csv', encoding='utf-8-sig')

# Первые 10 строк
print("Первые 10 строк заказов:")
print(df_orders_read.head(10))
print()

# Количество заказов каждого клиента
orders_per_client = df_orders_read['Клиент'].value_counts()
print("Количество заказов каждого клиента:")
print(orders_per_client)
print()

# Максимальная и минимальная суммы
max_sum = df_orders_read['Сумма'].max()
min_sum = df_orders_read['Сумма'].min()
print(f"Максимальная сумма заказа: {max_sum} руб.")
print(f"Минимальная сумма заказа: {min_sum} руб.")
print()

# Общая сумма всех заказов
total_sum = df_orders_read['Сумма'].sum()
print(f"Общая сумма всех заказов: {total_sum} руб.")
```

## Задание 3: DataFrame с продуктами питания


```python
# Создаю DataFrame с продуктами питания
products_data = {
    'Продукт': ['Яблоко', 'Бананы', 'Гречка', 'Рис', 'Куриная грудка', 
                'Лосось', 'Яйца', 'Творог', 'Овсянка', 'Авокадо',
                'Миндаль', 'Черника', 'Шоколад'],
    'Категория': ['Фрукты', 'Фрукты', 'Крупы', 'Крупы', 'Мясо',
                  'Рыба', 'Молочные', 'Молочные', 'Крупы', 'Фрукты',
                  'Орехи', 'Фрукты', 'Сладости'],
    'Калории': [52, 89, 137, 130, 165, 208, 155, 97, 389, 160, 579, 57, 546],
    'Белки': [0.3, 1.1, 4.2, 2.7, 31.0, 20.0, 13.0, 18.0, 16.9, 2.0, 21.2, 0.7, 7.8]
}

df_products = pd.DataFrame(products_data)

# Вывожу весь DataFrame
print("Исходный DataFrame с продуктами:")
print(df_products)
print()

# Продукты с калорийностью выше 300
high_calories = df_products[df_products['Калории'] > 300]
print("Продукты с калорийностью выше 300:")
print(high_calories)
print()

# Среднее количество белков по категориям
avg_proteins = df_products.groupby('Категория')['Белки'].mean()
print("Среднее количество белков по категориям:")
print(avg_proteins)
print()

# Сортировка по калорийности по убыванию
df_products_sorted = df_products.sort_values('Калории', ascending=False)
print("DataFrame, отсортированный по калорийности (убывание):")
print(df_products_sorted)
```

## Задание 4: DataFrame с сотрудниками и проектами


```python
# Создаю DataFrame с сотрудниками и проектами
employees_data = {
    'Имя': ['Иванов', 'Петров', 'Сидоров', 'Смирнов', 'Иванов',
            'Козлов', 'Петров', 'Новиков', 'Сидоров', 'Иванов'],
    'Проект': ['Проект А', 'Проект Б', 'Проект А', 'Проект В', 'Проект Б',
               'Проект А', 'Проект В', 'Проект Б', 'Проект А', 'Проект В'],
    'Часы': [120, 95, 150, 80, 130, 160, 110, 90, 140, 100]
}

df_employees = pd.DataFrame(employees_data)

# Вывожу исходный DataFrame
print("Исходный DataFrame с сотрудниками:")
print(df_employees)
print()

# Общее количество часов по каждому сотруднику
hours_per_employee = df_employees.groupby('Имя')['Часы'].sum().sort_values(ascending=False)
print("Общее количество часов по каждому сотруднику:")
print(hours_per_employee)
print()

# Общее количество часов по каждому проекту
hours_per_project = df_employees.groupby('Проект')['Часы'].sum().sort_values(ascending=False)
print("Общее количество часов по каждому проекту:")
print(hours_per_project)
print()

# Сотрудник, затративший больше всего часов
top_employee = hours_per_employee.index[0]
top_hours = hours_per_employee.iloc[0]
print(f"Сотрудник с наибольшим количеством часов: {top_employee} ({top_hours} часов)")
```

## Задание 5: DataFrame с продажами билетов


```python
# Создаю DataFrame с продажами билетов
tickets_data = {
    'Фильм': ['Интерстеллар', 'Интерстеллар', 'Интерстеллар', 'Матрица', 'Матрица',
              'Матрица', 'Гарри Поттер', 'Гарри Поттер', 'Гарри Поттер',
              'Аватар', 'Аватар', 'Аватар'],
    'Город': ['Москва', 'СПб', 'Казань', 'Москва', 'СПб', 'Казань',
              'Москва', 'СПб', 'Казань', 'Москва', 'СПб', 'Казань'],
    'Проданные билеты': [15000, 12000, 8000, 20000, 15000, 10000,
                         18000, 14000, 9000, 22000, 18000, 11000]
}

df_tickets = pd.DataFrame(tickets_data)

# Вывожу весь DataFrame
print("Исходный DataFrame с продажами билетов:")
print(df_tickets)
print()

# Общее количество проданных билетов по каждому фильму
tickets_per_film = df_tickets.groupby('Фильм')['Проданные билеты'].sum().sort_values(ascending=False)
print("Общее количество проданных билетов по каждому фильму:")
print(tickets_per_film)
print()

# Общее количество билетов по каждому городу
tickets_per_city = df_tickets.groupby('Город')['Проданные билеты'].sum().sort_values(ascending=False)
print("Общее количество билетов по каждому городу:")
print(tickets_per_city)
print()

# Фильм с наибольшим количеством продаж
top_film = tickets_per_film.index[0]
top_tickets = tickets_per_film.iloc[0]
print(f"Фильм с наибольшим количеством продаж: '{top_film}' ({top_tickets} билетов)")
```
